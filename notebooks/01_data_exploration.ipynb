```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

# Download the UCI credit default dataset
dataset = fetch_ucirepo(id=350)
```

```python
X = dataset.data.features.copy()
y = dataset.data.targets.iloc[:, 0].astype(int)

print("Dataset shape:", X.shape)
print("Target shape:", y.shape)
print("Default rate:", f"{y.mean():.2%}")

display(X.head())
```

```python

# Count customers by default outcome
default_counts = y.value_counts().sort_index()

plt.figure(figsize=(7, 4))
plt.bar(
    ["No default", "Default"],
    default_counts.values
)

plt.title("Distribution of Credit Card Defaults")
plt.ylabel("Number of Customers")
plt.xlabel("Default Status")
plt.show()

print("Default rate:", f"{y.mean():.2%}")
```
